```python
import os
import sys
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting aesthetics
sns.set_theme(style='whitegrid')
print('Environment and libraries initialized successfully.')
```

### 1. Corpus Inspection & Longitudinal Distribution
Loading the 2,000 papers collected from arXiv `cs.CL` and `cs.AI` across four distinct 2-year epochs.

```python
corpus_df = pd.read_csv('../data/processed/arxiv_nlp_corpus.csv')
print('Corpus Shape:', corpus_df.shape)
display(corpus_df.groupby('epoch')[['arxiv_id', 'year']].agg({'arxiv_id': 'count', 'year': ['min', 'max']}))
display(corpus_df.head(3)[['arxiv_id', 'title', 'epoch', 'primary_category']])
```

### 2. Lexical Dynamics: Emergence vs. Obsolescence Index
Tracking terms undergoing exponential growth vs. decay using the $Log_2$ Emergence Index.

```python
trends_df = pd.read_csv('../data/processed/lexical_emergence_decay_index.csv')
print('Top 10 Emergent Terms (Surging into Epoch 4):')
display(trends_df[trends_df['emergence_index'] > 1.5].head(10)[['term', 'norm_freq_e1', 'norm_freq_e4', 'emergence_index', 'category']])

print('\nTop 10 Decaying Terms (Collapsing by Epoch 4):')
display(trends_df[trends_df['emergence_index'] < -1.5].tail(10)[['term', 'norm_freq_e1', 'norm_freq_e4', 'emergence_index', 'category']])
```

### 3. Core Paradigm Markers Trajectories
Plotting normalized trajectories for key technical anchor keywords.

```python
traj_df = pd.read_csv('../data/processed/keyword_trajectories.csv')
pivot_traj = traj_df.pivot(index='epoch', columns='keyword', values='norm_freq_per_10k')

fig, ax = plt.subplots(figsize=(10, 5))
pivot_traj.plot(marker='o', linewidth=2.5, ax=ax)
plt.title('Longitudinal Keyword Frequencies per 10k Words (2018-2025)', fontweight='bold')
plt.ylabel('Normalized Frequency')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### 4. Topic Modeling & Longitudinal Evolution
Inspecting the 8 latent research topics discovered via Non-Negative Matrix Factorization (NMF).

```python
topic_dist = pd.read_csv('../data/processed/epoch_topic_distribution.csv')
display(topic_dist)

# NMF Topic Top Terms
nmf_terms = pd.read_csv('../data/processed/nmf_topic_terms.csv')
for t_id in range(1, 9):
    words = nmf_terms[nmf_terms['topic_id'] == f'Topic_{t_id}'].head(8)['term'].tolist()
    print(f'Topic {t_id}: {words}')
```

### 5. Semantic Geometry & Cluster Migration (Chi-Square Test)
Testing the statistical significance of document cluster migration across epochs.

```python
crosstab = pd.read_csv('../data/processed/cluster_epoch_crosstab.csv', index_col=0)
print('Cluster Share (%) per Epoch:')
display(crosstab)

from scipy.stats import chi2_contingency
chi2, p_val, dof, _ = chi2_contingency(crosstab)
print(f'Chi2 Statistic: {chi2:.2f}, p-value: {p_val:.4e}, Degrees of Freedom: {dof}')
print(f'Statistically Significant Paradigm Shift (p < 0.001): {p_val < 0.001}')
```
